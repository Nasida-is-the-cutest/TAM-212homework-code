```python
#HW5B.2. Point position given center of mass
from sympy import *

m1 = 9
m2 = 6
m3 = 2
r1 = Matrix([-7,  6,  0])
r3 = Matrix([ 8, -8,  0])
rC = Matrix([ 1, 10,  0])


a,b,c=symbols('a,b,c')
r2=Matrix([a,b,c])
eq=Eq(rC,(m1*r1+m2*r2+m3*r3)/(m1+m2+m3))
sol=solve(eq,(a,b,c),dict=True)
print(r2.subs(sol[0]))
```

```python
#HW5B.3. Moment of inertia about center of a segmented bar
from sympy import *

w = 3
l1 = 6
rho1 = 8
l2 = 6
rho2 = 9
l3 = 3
rho3 = 2


m1=l1*rho1*w
m2=l2*rho2*w
m3=l3*rho3*w

rc=(l1*m1/2+m2*(l1+l2/2)+m3*(l1+l2+l3/2))/(m1+m2+m3)
#print(rc) #以右边为起点

i1=m1*(l1**2+w**2)/12
i2=m2*(l2**2+w**2)/12
i3=m3*(l3**2+w**2)/12

d1=rc-l1/2
d2=rc-(l1+l2/2)
d3=l1+l2+l3/2-rc

I=i1+i2+i3+m1*d1**2+m2*d2**2+m3*d3**2
print(I)
```

```python
#HW5B.5. Moment of inertia about the center of an algebraic shape
import numpy as np
from sympy import *

x, y = symbols('x, y')

rho = 6
y1 = -2*x**2 - 3
y2 = x**2 + 4*x + 7
x1 = -1
x2 = 0

f=(x**2+y**2)*rho
I=integrate(f,(y,y1,y2),(x,x1,x2)).evalf()#绕原点的I

fm=rho
m=integrate(fm,(y,y1,y2),(x,x1,x2))#质量

f1=rho*x/m
mcx=integrate(f1,(y,y1,y2),(x,x1,x2)).evalf()
f2=y*rho/m
mcy=integrate(f2,(y,y1,y2),(x,x1,x2)).evalf()
#计算质心的位置，双重积分
Ic=I-m*(mcx**2+mcy**2)
print(Ic)
```

```python
#HW5B.6. Slider-crank kinetics
from sympy import *

m1 = 2.000
m2 = 6.000
I1 = 2.000
I2 = 34.000
rOP = Matrix([-1, -1,  0])
rPQ = Matrix([ 8, -2,  0])
omega1 = Matrix([0.000, 0.000, -16.000])
omega2 = Matrix([ 0,  0, -2])
vP = Matrix([-16.000, 16.000, 0.000])
vQ = Matrix([-20.000, 0.000, 0.000])
alpha2 = Matrix([0.000, 0.000, 1.000])
aQ = Matrix([498.000, 0.000, 0.000])

aP=aQ-alpha2.cross(rPQ)-omega2.cross(omega2.cross(rPQ))
a=symbols('a')
alpha1=Matrix([0,0,a])
eq=Eq(aP,alpha1.cross(rOP)+omega1.cross(omega1.cross(rOP)))
sol=solve(eq,a,dict=True)
alpha1=alpha1.subs(sol[0])
#print(alpha1)
m1=alpha1.norm()*I1

rPC = rPQ / 2  # 质心C在连杆中点，相对P点的位置
aC = aP + alpha2.cross(rPC) + omega2.cross(omega2.cross(rPC))
FPx, FPy, FQy = symbols('FPx FPy FQy')
# 牛顿第二定律：F_P + F_Q = m2 aC
eq1 = Eq(FPx, m2 * aC[0])  # F_Qx=0，故FPx直接由x方向平衡得到
eq2 = Eq(FPy + FQy, m2 * aC[1])
# 绕质心C的力矩平衡：r_CP×F_P + r_CQ×F_Q = I2 alpha2
rCP = -rPC  # 相对质心C，P点的位置矢量
rCQ = rPQ/2 # 相对质心C，Q点的位置矢量
torque_P = rCP.cross(Matrix([FPx, FPy, 0]))
torque_Q = rCQ.cross(Matrix([0, FQy, 0]))  # F_Qx=0
eq3 = Eq(torque_P[2] + torque_Q[2], I2 * alpha2[2])
sol_F = solve((eq1, eq2, eq3), (FPx, FPy, FQy), dict=True)[0]
F_P = Matrix([sol_F[FPx], sol_F[FPy], 0])
m=rOP.cross(F_P).norm()+m1
print(m)

```

```python
#HW5B.6. Slider-crank kinetics
from sympy import *

m1 = 4.000
m2 = 3.000
I1 = 4.000
I2 = 17.000
rOP = Matrix([ 1, -1,  0])
rPQ = Matrix([2, 8, 0])
omega1 = Matrix([0.000, 0.000, -16.000])
omega2 = Matrix([ 0,  0, -2])
vP = Matrix([-16.000, -16.000, 0.000])
vQ = Matrix([0.000, -20.000, 0.000])
alpha1 = Matrix([0.000, 0.000, 280.000])
alpha2 = Matrix([0.000, 0.000, 2.000])
aP = Matrix([24.000, 536.000, 0.000])

# 1. 计算质心加速度
rPC = rPQ / 2
aC = aP + alpha2.cross(rPC) + omega2.cross(omega2.cross(rPC))

# 2. 定义未知数与方程
FPx, FPy, FQx = symbols('FPx FPy FQx')
FQy = 0  # 光滑竖直槽，y方向无约束力

fx_eq = Eq(FPx + FQx, m2 * aC[0])
fy_eq = Eq(FPy + FQy, m2 * aC[1])

rCP = -rPC
rCQ = rPC
torque_P = rCP.cross(Matrix([FPx, FPy, 0]))
torque_Q = rCQ.cross(Matrix([FQx, FQy, 0]))
moment_eq = Eq(torque_P[2] + torque_Q[2], I2 * alpha2[2])

sol = solve((fx_eq, fy_eq, moment_eq), (FPx, FPy, FQx), dict=True)[0]

print(f"F_Q = {sol[FQx]} î + {FQy} ĵ N")

```
