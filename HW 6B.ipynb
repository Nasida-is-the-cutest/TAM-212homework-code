```python
#HW6B.3. Acceleration on a body rolling on a curved surface
from sympy import *

rho = 52
omega = Matrix([0, 0, 2])
alpha = Matrix([ 0,  0, -2])
rCQ = Matrix([-8, -4,  0])


r=rCQ.norm()

rCP=Matrix([r, 0,  0])  #会随图变化

vc=omega.cross(rCP)
ac_t=r*alpha.norm() #切向加速度的大小a=r*alpha
ac_n=vc.norm()**2/(rho-r)
ac=Matrix([-ac_n, ac_t, 0]) #切向加速度的方向需要与速度方向相反（减速运动），随图变化
aq=ac+alpha.cross(rCQ)+omega.cross(omega.cross(rCQ))
print(aq.evalf())
```

```python
#HW6B.5. Rolling body offset-point acceleration
from sympy import *

# 已知条件
omega = Matrix([0, 0, 6])
aQ = Matrix([1, -5, 0])

# 未知量：角加速度alpha（z分量）和半径R
alpha, R = symbols('alpha R', real=True)

# 位置矢量：从C到Q（左侧，x负方向）
rCQ = Matrix([-R, 0, 0])

# 质心加速度（纯滚动条件：aC = -alpha*R * i）
aC = Matrix([-alpha*R, 0, 0])###############################

# 切向加速度：alpha_vec × rCQ
alpha_vec = Matrix([0, 0, alpha])
tangential_acc = alpha_vec.cross(rCQ)

# 法向加速度：omega × (omega × rCQ)
normal_acc = omega.cross(omega.cross(rCQ))

# 总加速度合成
aQ_calc = aC + tangential_acc + normal_acc

# 建立方程并求解
eq1 = Eq(aQ_calc[0], aQ[0])  # x分量匹配
eq2 = Eq(aQ_calc[1], aQ[1])  # y分量匹配
solution = solve((eq1, eq2), (alpha, R))

print("角加速度α =", solution[0][0], "rad/s²")
print("半径R =", solution[0][1], "m")
```

```python
#HW6B.5. Rolling body offset-point acceleration
from sympy import *

r = 7
aQ = Matrix([-2, -3,  0])
w, alpha = symbols('w alpha', real=True)
rCQ = Matrix([r, 0, 0]) 
ac=Matrix([-alpha*r, 0, 0])
alpha2=Matrix([0, 0, alpha])
omega = Matrix([0, 0, w])
aq=ac+alpha2.cross(rCQ)+omega.cross(omega.cross(rCQ))
eq=Eq(aq, aQ)
solution = solve(eq, (w, alpha))
print("角速度w =", solution[0][0], "rad/s")
print("角加速度alpha =", solution[0][1], "rad/s²")
```
